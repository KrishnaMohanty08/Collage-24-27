Pandas + Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample DataFrame
data = {'Values': [1, 2, 3, 4, 5]}
df = pd.DataFrame(data)

# Additional Pandas operations
print("DataFrame info:")
print(df.info())
print("\nDataFrame shape:", df.shape)
print("\nDataFrame column names:", df.columns.tolist())

# Sorting and filtering
print("\nFiltered values (> 2):")
print(df[df['Values'] > 2])

# Adding new columns
df['Squared'] = df['Values'] ** 2
df['Doubled'] = df['Values'] * 2
print("\nDataFrame with new columns:")
print(df)

# Grouping and aggregation
print("\nSum of values:", df['Values'].sum())
print("Product of values:", df['Values'].prod())

# Basic Matplotlib plotting with enhancements
plt.figure(figsize=(14, 10))

# Line plot
plt.subplot(2, 3, 1)
plt.plot(df['Values'], marker='o', linewidth=2, markersize=8, color='blue')
plt.title('Line Plot')
plt.xlabel('Index')
plt.ylabel('Values')
plt.grid(True)

# Bar plot
plt.subplot(2, 3, 2)
plt.bar(range(len(df)), df['Values'], color='green', alpha=0.7)
plt.title('Bar Plot')
plt.xlabel('Index')
plt.ylabel('Values')

# Scatter plot
plt.subplot(2, 3, 3)
plt.scatter(df['Values'], df['Squared'], color='red', s=100)
plt.title('Scatter Plot')
plt.xlabel('Values')
plt.ylabel('Squared Values')

# Multiple line plot
plt.subplot(2, 3, 4)
plt.plot(df['Values'], label='Original', marker='o')
plt.plot(df['Doubled'], label='Doubled', marker='s')
plt.legend()
plt.title('Multiple Lines')
plt.xlabel('Index')
plt.ylabel('Values')
plt.grid(True)

# Histogram
plt.subplot(2, 3, 5)
plt.hist(df['Values'], bins=5, color='purple', alpha=0.7, edgecolor='black')
plt.title('Histogram')
plt.xlabel('Values')
plt.ylabel('Frequency')

# Box plot
plt.subplot(2, 3, 6)
plt.boxplot([df['Values'], df['Doubled']], labels=['Original', 'Doubled'])
plt.title('Box Plot')
plt.ylabel('Values')

plt.tight_layout()
plt.show()
```
