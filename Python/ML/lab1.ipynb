```python
import numpy as np
import pandas as pd

# rawset

data = np.array([
    [160, 55],
    [165, 60],
    [170, 65],
    [175, 70],
    [180, 72]
])

df = pd.DataFrame(data, columns=["Height", "Weight"])
print("===== Dataset =====")
print(df)

# Convert to numpy arrays
X1 = df["Height"].values
X2 = df["Weight"].values

n = len(X1)

# ==========================
# Step 1 — Mean of Variables
# ==========================
mean_X1 = np.mean(X1)
mean_X2 = np.mean(X2)

print("\nMean Height:", mean_X1)
print("Mean Weight:", mean_X2)

# ==========================
# Step 2 — Manual Covariance
# ==========================
# Cov(X,Y) = Σ((Xi - meanX)(Yi - meanY)) / (n - 1)
cov_X1_X1 = np.sum((X1 - mean_X1) * (X1 - mean_X1)) / (n - 1)
cov_X2_X2 = np.sum((X2 - mean_X2) * (X2 - mean_X2)) / (n - 1)
cov_X1_X2 = np.sum((X1 - mean_X1) * (X2 - mean_X2)) / (n - 1)

cov_matrix = np.array([
    [cov_X1_X1, cov_X1_X2],
    [cov_X1_X2, cov_X2_X2]
])

print("\n===== Covariance Matrix (Manual) =====")
print(cov_matrix)

# ==========================
# Step 3 — Manual Correlation
# ==========================
# Corr = cov / (std1 * std2)
std_X1 = np.sqrt(cov_X1_X1)
std_X2 = np.sqrt(cov_X2_X2)

corr_X1_X2 = cov_X1_X2 / (std_X1 * std_X2)

corr_matrix = np.array([
    [1,            corr_X1_X2],
    [corr_X1_X2,   1]
])

print("\n===== Correlation Matrix (Manual) =====")
print(corr_matrix)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data: [Height (cm), Weight (kg)]
data = np.array([
    [160, 55],
    [165, 60],
    [170, 65],
    [175, 70],
    [180, 72]
])

# Split into height and weight
height = data[:, 0]
weight = data[:, 1]

# Create a dual boxplot
plt.boxplot([height, weight], labels=['Height (cm)', 'Weight (kg)'])
plt.title("Dual Boxplot: Height vs Weight")
plt.ylabel("Value")
plt.show()
```
